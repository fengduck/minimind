```python
from transformers import AutoTokenizer
messages = [
    {"role": "user", "content": "你好"},
    {"role": "assistant", "content": "你好，有什么我可以帮你的吗？"}
]
tokenizer = AutoTokenizer.from_pretrained('./model')
text = tokenizer.apply_chat_template(messages, tokenize=False)

bos_id = tokenizer('<|im_start|>assistant')
eos_id = tokenizer('<|im_end|>', add_special_tokens=False)
print(f'text = {text}')
print(f'bos_id = {bos_id}')
print(f'eos_id = {eos_id}')
print(tokenizer(text))
print(tokenizer(text)['input_ids'])
tokens = tokenizer("\n")          # 把换行符交给 tokenizer
print("\\n =", tokens)           # 在输出里显示字面量 \n
# 或更可读：
print("tokenizer('\\n') =", tokens)
assistant = tokenizer('assistant')
print(assistant)
user = tokenizer('user')
print(user)
# 将input_ids转回文本
# 对于 transformers 的 fast tokenizer，调用 tokenizer(...) 返回的是一个 BatchEncoding 对象。
# BatchEncoding 既可以像字典一样访问（bos_id['input_ids']），也有属性访问方式（bos_id.input_ids）。
print(tokenizer.decode(bos_id['input_ids'] + [5134, 270, 5183, 5213, 1086, 2155, 2235, 814] + eos_id.input_ids))
```

```python
# 测试字典类型的取值方法
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(my_dict['a'])
print(my_dict.get('b'))
print(my_dict.get('d', 4))  # 如果key不存在，返回默认值4
# 字典不允许这样取值
# print(my_dict.b)
l = [(k,v) for k,v in my_dict.items() if v>1 ]  # 生成新的元组
print(l)
for k,v in enumerate(l, 11):
    print(k,v)

l_dict = {k:v for k,v in my_dict.items() if v>1}  # 生成新的字典
print(l_dict)
```
